```python
x = df.iloc[:, [0,1,2,3]].values
```

```python
plt.scatter(x[:,0], x[:,1], label='True Position')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('Sepal Length vs Sepal Width Grafic')
plt.show()
```

```python
plt.scatter(x[:,2], x[:,3], label='True Position')
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.title('Petal Length vs Petal Width Grafic')
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
x_scaled = scaler.fit_transform(x)
pd.DataFrame(x_scaled, columns=["a", "b", "c", "d"])
```

```python
kmeans = KMeans(n_clusters = 4, random_state = 123)
kmeans.fit(x)
print(kmeans)
```

```python
print(kmeans.cluster_centers_)
```

```python
print(np.unique(kmeans.labels_))
```

```python
plt.scatter(x_scaled[:,0], x_scaled[:,1], c=kmeans.labels_, cmap="rainbow")
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('Cluster Grafic')
plt.show()
```

```python
df["cluster"] = kmeans.labels_
output = plt.scatter(x_scaled[:,0], x_scaled[:,1], c = df.cluster, marker = "o", alpha = 1, )
centers = kmeans.cluster_centers_
plt.scatter(centers[:,0], centers[:,1], c='green', alpha=1, marker="s")
plt.axis([0,1.2,0,1.2])
plt.title("KMeans Clustering result")
plt.colorbar(output)
plt.show()
```

```python
Error = []
for i in range(1,11) :
  kmeans = KMeans(n_clusters = i).fit(x)
  kmeans.fit(x)
  Error.append(kmeans.inertia_)
plt.plot(range(1,11), Error)
plt.title('Elbow Method')
plt.xlabel('No of clusters')
plt.ylabel('Error')
plt.show()
```

```python
sepal_length =  3
sepal_width =  2
petal_length =  1
petal_width =  1

x_try = [[sepal_length, sepal_width, petal_length, petal_width]]
x_try = scaler.transform(x_try)

y_pred = kmeans.predict(x_try)

print("Hasil prediksi: %s" % y_pred[0])
```
